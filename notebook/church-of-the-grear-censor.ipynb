```python
import os
import sys
import time
from pathlib import Path
from PIL import Image
import numpy as np


class ImageCensor:
    def __init__(self):
        self.default_size = None

    def print_slow(self, text, delay=0.05, line_delay=0.05):
        for line in text.split('\n'):
            for c in line:
                sys.stdout.write(c)
                sys.stdout.flush()
                time.sleep(delay)
            print()
            time.sleep(line_delay)

    def censor_object(self, input_path, color=(0, 0, 0), factor_pixelado=8, umbral=240):
        try:
            input_path = Path(input_path)

            # Crear carpeta de salida
            output_dir = Path("output")
            output_dir.mkdir(exist_ok=True)

            output_path = output_dir / f"{input_path.stem}_censored.png"

            with Image.open(input_path) as imagen:
                # Convertir siempre a RGB para soportar colores
                imagen = imagen.convert("RGB")
                img_array = np.array(imagen)

                # Generar versión en escala de grises SOLO para la máscara
                img_gray = np.mean(img_array, axis=2).astype(np.uint8)

                # Detectar fondo blanco y objeto
                mascara_fondo = img_gray > umbral
                mascara_objeto = ~mascara_fondo

                altura, ancho, _ = img_array.shape

                # Aplicar el color a la imagen RGB original
                for i in range(0, altura, factor_pixelado):
                    for j in range(0, ancho, factor_pixelado):

                        mascara_bloque = mascara_objeto[
                            i:i + factor_pixelado,
                            j:j + factor_pixelado
                        ]

                        if np.any(mascara_bloque):
                            # Asignar la tupla de color RGB a los bloques seleccionados
                            img_array[
                                i:i + factor_pixelado,
                                j:j + factor_pixelado
                            ] = color

                # Guardar resultado final en color
                imagen_censurada = Image.fromarray(img_array)
                imagen_censurada.save(output_path)

                print(
                    f"\nHappy to help! Take a look at"
                    f"\n{output_path.resolve()}"
                )

                return True

        except Exception as e:
            print(f"\nYour censorship has been a fraud! Please, check yourself... {e}")
            return False

    def get_file_name(self, prompt):
        while True:
            name = input(prompt).strip()

            if name.lower() == "exit":
                return None

            if not os.path.isfile(name):
                print(f"File '{name}' not found. Try again.")
                continue

            return name

    def get_color(self, prompt):
        """Solicita un color al usuario y devuelve una tupla RGB."""
        while True:
            color_input = input(prompt).strip().lower()
            if color_input == "exit":
                return None
            
            colors = {
                "1": (49, 25, 205), #azul palo de helado
                "2": (225, 16, 72), #magenta
                "3": (203, 203, 15), #amarillo
                "4": (158, 184, 44), #lima
                '5': (181, 101, 43), #naranja
                "6": (20, 19, 30), #negro lavado
            }
            if color_input in colors:
                return colors[color_input]
            elif "," in color_input:
                try:
                    r, g, b = map(int, color_input.split(","))
                    if 0 <= r <= 255 and 0 <= g <= 255 and 0 <= b <= 255:
                        return (r, g, b)
                    else:
                        print("RGB values must be between 0, 255.")
                        continue
                except ValueError:
                    print("If you're using 'R,G,B', use this format: 255,0,0').")
                    continue
            else:
                print("Try a color available: red, green, blue, black, pink or create your own using 'R,G,B'.")

    def run(self):

        print("""
        
                ░█▀█░█▄█░█▀█░▀█▀░█▀█░░░█░█░█▀█░█░█░█▀▀░░░█▀▀░█▀▀░▀█▀
                ░█░█░█░█░█░█░░█░░█▀█░░░█░█░█░█░█░█░▀▀█░░░█▀▀░▀▀█░░█░
                ░▀▀▀░▀░▀░▀░▀░▀▀▀░▀░▀░░░▀▀▀░▀░▀░▀▀▀░▀▀▀░░░▀▀▀░▀▀▀░░▀░
                                p r e s e n t s
        """)

        self.print_slow(
            """
                              INSTANT CENSORSHIP
            """,
            delay=0.01,
            line_delay=0.05
        )

        ascii_art = r"""
                .                  .             * ·                  *
        ·           .                                               .
    .                     .                    .
  .                                 * .      .         *               #
                              =        .            =
                .          <^\O|^                 ^|O/^>          *
         .               * \/  |/                 \|  \|
                            /__|      .  '  .      |__\                 #
                =           ll     .    |          ll
                            ##          #           ################
             <^\O|^     ¡__/||       '  |  '       ||\__¡     ######
             \/  |/   ¡_/_| ||  .  '  \ | / '  #   || |_\_¡   \|  \
      .       /__|  ######|_||      -== + ==-      ||_||_||_¡  |__\
            ##################      #########      ##########  ### ########
            ################################################################
            $$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$$############
            ##################$#############################################
            ################################################################
            ################################################################
            C;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;;C
            ######################                     #####################
            #####################     W E L C O M E     ####################
            ###################                           ##################
            ##################             to              #################
            ##################                             #################
            ##################             the             #################
            ==================                             =================
            ##################         C H U R C H         #################
            ##################                             #################
            ##################             of              #################
            ##################                             #################
            ##################            T H E            #################
            ##################                             #################
            ;;;;;;;;;;;;;;;;;;          G R E A T          ;;;;;;;;;;;;;;;;;
            ;;;;;;;;;;;;;;;;;;                             ;;;;;;;;;;;;;;;;;
            ;;;;;;;;;;;;;;;;;;         C E N S O R         ;;;;;;;;;;;;;;;;;
        """

        self.print_slow(ascii_art, delay=0.01, line_delay=0.01)

        self.print_slow(
            "Please:\nPresent into The Great Censor the image you want to put under a veil.\n\n"
            "Take all the time you need.\n\nTime is al we have",
            delay=0.05,
            line_delay=0.8
        )

        while True:
            input_path = self.get_file_name(
                "\n> Type the name of your image: "
            )

            if input_path is None:
                break

            chosen_color = self.get_color('\n> Choose a color for your Summer of Censorship era (1, 2, 3, 4, 5, 6)')

            if chosen_color is None:
                break

            # Se llama a la función UNA SOLA VEZ pasando el color
            if self.censor_object(input_path, color=chosen_color):
                print("\nYou've been granted Instant Censorship! Congratulations!")

            again = input(
                "> Do you have more offerings to The Great Censor (Y/N): "
            ).strip().lower()

            if again != "y":
                self.print_slow(
                    "\nHappy to help.\n\nTime is all we have.\n\n"
                    "Use it wisely... or not.\n\n"
                    "It's not like we are counting.",
                    delay=0.05,
                    line_delay=0.8
                )
                break


if __name__ == "__main__":
    censor = ImageCensor()
    censor.run()
```
