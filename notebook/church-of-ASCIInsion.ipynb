```python
from PIL import Image
import numpy as np
import os
import time
import sys

class ASCIIfy:
    def __init__(self):
        self.ascii_chars = [' ', '.', ':', ';', '-', '+', 'x', '░', '▒', '▓', '█']

    def print_slow(self, text, delay=0.05, line_delay=0.5):
        for line in text.split('\n'):
            for c in line:
                sys.stdout.write(c)
                sys.stdout.flush()
                time.sleep(delay)
            print()
            time.sleep(line_delay)

    def get_file_name(self, prompt, default_extension=".png"):
        while True:
            name = input(prompt).strip()
            if name.lower() == 'exit':
                return None
            if not os.path.splitext(name)[1]:
                name += default_extension
            if default_extension in ('.png', '.jpg', '.jpeg') and not os.path.exists(name):
                print(f"File '{name}' not found. Try again.")
                continue
            return name

    def imagen_a_ascii(self, imagen_path, ancho_salida=30, alto_salida=None):
        imagen = Image.open(imagen_path).convert('L')  # Convertir a escala de grises
        ancho_original, alto_original = imagen.size

        if alto_salida is None:
            # Calcular altura proporcional si no se especifica
            ratio = alto_original / ancho_original
            alto_salida = int(ancho_salida * ratio)

        # Redimensionar la imagen al tamaño deseado
        imagen = imagen.resize((ancho_salida, alto_salida))
        pixeles = np.array(imagen)

        # Normalizar los píxeles al rango de caracteres ASCII
        pixeles = np.interp(pixeles, (pixeles.min(), pixeles.max()), (0, len(self.ascii_chars)-1))

        ascii_str = ""
        for fila in pixeles:
            for pixel in fila:
                ascii_str += self.ascii_chars[int(pixel)]
            ascii_str += "\n"
        return ascii_str

    def run(self):
        self.print_slow("""
                         C H U R C H  o f  A S C I I n s i o n
        """, delay=0.01, line_delay=0.05)

        ascii_art = r"""
    .                     .                    .
  .                                 *                .      .         *
                             =        .            =                   
                .         <^\O|^                 ^|O/^>          *
         .         *      \/  |/                 \|  \|                .        .     *
                           /__|      .  '  .      |__\     *
                =           ll     .    |          ll           =                     
             <^\O|^     ¡__/||       '  |  '       ||\__¡     ^|O/^>      .         *  ·
             \/  |/   ¡_/_| ||  .  '  \ | / '  .   || |_\_¡   \|  \/
      .       /__|  ¡_||_||_||      -== + ==-      ||_||_||_¡  |__\    .       ·
             ..ll   //-|--|-||  '  .  / . \  .  '  ||-|--|-\\   ll..
            (____) /|  |  | ||       .  |  .       || |  |  |\ (____)
             |  |¡//|  |  | ||    '     |     '    || |  |  |\\¡| -|          .  ·
             | -|¡  |  |  | ||       '  .  '       || |  |  | \¡|  |
             |  ||  |  |  | ||                     || |  |  |  || -|
        ¡_¡__|- ||  |  |  | ||   *      +          || |  |  |  ||  |__¡_¡
        ¡_|__|__||__|__|__|_||          A          ||_|__|__|__||__|__|_¡
        ·-|--|--||--|--|--|-||       _ /_\__  *    ||-|--|--|--||--|--|-·
          |  |  ||  |  |  | ||      /|-'O'-|\      || |  |  |  || -|  |
          |  |- ||  |  |  | ||     _||-----||_     || |  |  |  ||  |  |
          |  |  ||  |  |  | ||     /||=====||\     || |  |  |  ||- |  |
          |  | -||  |  |  | ||  __|I_I_[_]_I_I|__  || |  |  |  ||  |  |
         _|__|__||__|__|__|_||:::::::::::::::::::::||_|__|__|__||__|__|_
    www  -|--|--||--|--|--|-||::::::::::::::::www::||-|--|--|--||--|--|-  www
   wwwww =|==|==||==|==|==|=||::::::::::::www::::::||=|==|==|==||==|==|= wwwwwwwwwww  www
wwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwwww
        """
        self.print_slow(ascii_art, delay=0.01, line_delay=0)

        self.print_slow(
            "               W E L C O M E   T O   T H E   C O N V E R S I O N   P R O G R A M",
            delay=0.05, line_delay=0.10
        )

        self.print_slow("Please:\nGather in my destiny (folder) the chosen image you want to disintegrate into ASCII.\n\nTake all the time you need.\n\nTime is all we have.",
            delay=0.05, line_delay=0.8)

        while True:
            input_path = self.get_file_name(
                "\n> What is the name of the one you wish to join the Ascension? (e.g., image.png): ",
                '.png')
            if input_path is None:
                break

            ancho_input = input('> Do you want a width (in number of characters)? Press Enter for default (30): ')
            ancho = int(ancho_input) if ancho_input else 30
            alto_input = input('> Do you want a height (leave empty for proportional calculation)? Press Enter to skip: ')
            alto = int(alto_input) if alto_input else None

            ascii_art = self.imagen_a_ascii(input_path, ancho, alto)
            self.print_slow(ascii_art, delay=0.01, line_delay=0.1)

            output_path = self.get_file_name(
                "\n> Do you want to save the ASCII art to a file? Enter the filename (or press Enter to skip): ",
                '.txt')
            if output_path:
                with open(output_path, 'w') as f:
                    f.write(ascii_art)
                print(f"ASCII art saved to: {os.path.abspath(output_path)}")

            onemoretime = input("> Do you wish to join another icon? (Y/N): ").strip().lower()
            if onemoretime == 'n':
                self.print_slow("\nTime is all we have.\n\nUse it wisely... or don't.\n\nIt's not like we're counting.", delay=0.05, line_delay=0.8)
                break

if __name__ == "__main__":
    asciify = ASCIIfy()
    asciify.run()
```
