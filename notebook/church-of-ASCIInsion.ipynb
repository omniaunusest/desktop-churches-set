```python
import os
import sys
from PIL import Image
import numpy as np
import time

#preguntar por el path?
os.chdir(r"C:\Users\OMNA\Desktop")

def print_slow(text, delay=0.001, line_delay=0):
    for line in text.split('\n'):
        for c in line:
            sys.stdout.write(c)
            sys.stdout.flush()
            time.sleep(delay)
        print()
        time.sleep(line_delay)

'''ascii_chars = [" ", "░", "▒", "▓", "█"]'''
ascii_chars = [' ', '.', ':', ';', '-', '+', 'x']

def imagen_a_ascii(imagen_path, ancho_salida=30, alto_salida=None):
    imagen = Image.open(imagen_path).convert('L')
    ancho_original, alto_original = imagen.size

    if alto_salida is None:
        ratio = alto_original / ancho_original
        alto_salida = int(ancho_salida * ratio)

    imagen = imagen.resize((ancho_salida, alto_salida))
    pixeles = np.array(imagen)
    pixeles = np.interp(pixeles, (pixeles.min(), pixeles.max()), (0, len(ascii_chars)-1))

    ascii_str = ""
    for fila in pixeles:
        for pixel in fila:
            ascii_str += ascii_chars[int(pixel)]
        ascii_str += "\n"
    return ascii_str

def get_file_name(prompt, default_extension=".png"):
    while True:
        name = input(prompt).strip()
        if name.lower() == 'exit':
            return None
        if not os.path.splitext(name)[1]:
            name += default_extension
        if not os.path.exists(name):
            print(f"File '{name}' not found. Try again or type 'exit' to quit.")
            continue
        return name

def run():
    entra = get_file_name('> Nombre de la imagen: ')
    if entra is None:
        return
    ancho = input('> Desea un ancho (en número de caracteres): ')
    ancho = int(ancho) if ancho else 30
    alto_input = input('> Desea una altura (dejar vacío para calcular proporcionalmente): ')
    alto = int(alto_input) if alto_input else None
    ascii_art = imagen_a_ascii(entra, ancho, alto)
    print_slow(ascii_art)

run()

```
